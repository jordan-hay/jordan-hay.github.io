```python
import sympy as sp

# Given
V_final = sp.Rational(500, 1000)   # L
C_final = sp.Rational(20, 100)     # M
C_stock = sp.Integer(1)            # M
C_HCl = sp.Integer(2)              # M

pH_target = sp.Float(6.20)
pH_stock = sp.Float(11.30)
pKa1 = sp.Float(6.35)
pKa2 = sp.Float(10.33)

# Total carbonate needed
n_total = C_final * V_final
V_stock = n_total / C_stock

# Target ratio: HCO3-/H2CO3
r_target = 10**(pH_target - pKa1)

H2CO3_f = n_total / (1 + r_target)
HCO3_f = r_target * H2CO3_f

# Initial stock ratio: CO3^2-/HCO3-
r_stock = 10**(pH_stock - pKa2)

HCO3_i = n_total / (1 + r_stock)
CO3_i = r_stock * HCO3_i

# H+ accounting:
# CO3^2- -> HCO3- uses 1 H+
# HCO3- -> H2CO3 uses 1 H+
# Equivalent compact formula:
# H+ needed = final proton equivalents - initial proton equivalents
H_initial = HCO3_i + 0*CO3_i
H_final = HCO3_f + 2*H2CO3_f

n_HCl = H_final - H_initial
V_HCl = n_HCl / C_HCl

print("Total carbonate mol:", sp.N(n_total, 5))
print("Carbonate stock volume L:", sp.N(V_stock, 5))
print("Carbonate stock volume mL:", sp.N(1000*V_stock, 5))

print("Final H2CO3 mol:", sp.N(H2CO3_f, 5))
print("Final HCO3- mol:", sp.N(HCO3_f, 5))

print("Initial HCO3- mol:", sp.N(HCO3_i, 5))
print("Initial CO3^2- mol:", sp.N(CO3_i, 5))

print("HCl mol needed:", sp.N(n_HCl, 5))
print("2.0 M HCl volume L:", sp.N(V_HCl, 5))
print("2.0 M HCl volume mL:", sp.N(1000*V_HCl, 5))
```

```python
import numpy as np
import plotly.graph_objects as go

# Henderson–Hasselbalch parameters
pKa = 6.35

# pH range
pH = np.linspace(3, 10, 400)

# [A-]/[HA] ratio from Henderson–Hasselbalch
ratio = 10**(pH - pKa)

# Create figure
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=pH,
        y=ratio,
        mode='lines',
        name='[HCO₃⁻]/[H₂CO₃]'
    )
)

# Add vertical line at pKa
fig.add_vline(
    x=pKa,
    line_dash="dash",
    annotation_text=f"pKa = {pKa}",
    annotation_position="top"
)

# Layout
fig.update_layout(
    title="Henderson–Hasselbalch Relationship for the Carbonic Acid Buffer",
    xaxis_title="pH",
    yaxis_title="[HCO₃⁻] / [H₂CO₃]",
    yaxis_type="log",
    template="plotly_white"
)

fig.show()
```
