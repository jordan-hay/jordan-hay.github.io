```python
import math

C = 0.10
pKa_arg = 12.5

Ka = 10**(-pKa_arg)
Kb = 1e-14 / Ka

# Kb = x^2 / (C - x)
x = (-Kb + math.sqrt(Kb**2 + 4 * Kb * C)) / 2
#x = math.sqrt(Kb * C)

pOH = -math.log10(x)
pH = 14 - pOH

print(f"Kb = {Kb:.6e}")
print(f"[OH-] = {x:.6e} M")
print(f"pOH = {pOH:.6f}")
print(f"pH = {pH:.6f}")
```

```python
pKas = [3.6, 3.9, 6.0, 7.4, 8.3, 10.1, 10.5, 12.5]
C = 0.10

def f(pH):
    OH = 10**(pH - 14)
    bound_H = C * sum(
        1 / (1 + 10**(pH - pKa))
        for pKa in pKas
    )
    return OH - bound_H

low, high = 0.0, 14.0

for _ in range(100):
    mid = (low + high) / 2

    if f(low) * f(mid) <= 0:
        high = mid
    else:
        low = mid

pH = (low + high) / 2
OH = 10**(pH - 14)

print(f"pH = {pH:.6f}")
print(f"[OH-] = {OH:.6e} M")
```

```python
from scipy.optimize import fsolve

pKas = [3.6, 3.9, 6.0, 7.4, 8.3, 10.1, 10.5, 12.5]
C = 0.10

def equation(pH):
    OH = 10**(pH - 14)
    bound_H = C * sum(
        1 / (1 + 10**(pH - pKa))
        for pKa in pKas
    )
    return OH - bound_H

pH = fsolve(equation, 12.6)[0]
OH = 10**(pH - 14)

print(f"pH = {pH:.6f}")
print(f"[OH-] = {OH:.6e} M")
```
